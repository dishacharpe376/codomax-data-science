```python
import pandas as pd
import numpy as np

# Create sample dataset with missing values and duplicates
data = {
    "Name": ["Disha", "Rahul", "Priya", "Aman", "Disha"],
    "Age": [19, 20, np.nan, 21, 19],
    "City": ["Bhopal", "Indore", "Ujjain", np.nan, "Bhopal"],
    "Marks": [90, 85, 95, np.nan, 90]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["City"] = df["City"].fillna("Unknown")
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nCleaned Dataset:")
print(df)

# Check data types
print("\nData Types:")
print(df.dtypes)
```
