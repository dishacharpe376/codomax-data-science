```python
import pandas as pd

# Sample Dataset
data = {
    "Name": ["Disha", "Rahul", "Priya", "Aman", "Sneha"],
    "Age": [19, 20, 18, 21, 19],
    "Marks": [90, 85, 95, 88, 92]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("Dataset:")
print(df)

# Save DataFrame to CSV
df.to_csv("students_data.csv", index=False)

print("\nCSV file has been created successfully!")
```
