```python
import pandas as pd

# Sample Dataset
data = {
    "Age": [18, 19, 20, 21, 22],
    "Marks": [78, 85, 90, 88, 95],
    "Study_Hours": [2, 3, 4, 4, 5]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("Dataset:")
print(df)

# Correlation Matrix
print("\nCorrelation Matrix:")
print(df.corr())

# Average Values
print("\nAverage Age:", df["Age"].mean())
print("Average Marks:", df["Marks"].mean())
print("Average Study Hours:", df["Study_Hours"].mean())
```
