```python
import pandas as pd

# Sample Dataset
data = {
    "Name": ["Disha", "Rahul", "Priya", "Aman", "Sneha"],
    "Marks": [90, 85, 95, 88, 92]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Data Analysis
print("\nTotal Marks:", df["Marks"].sum())
print("Average Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
print("Number of Students:", df["Marks"].count())
```
