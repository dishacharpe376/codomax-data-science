```python
import pandas as pd

# Sample Dataset
data = {
    "Name": ["Disha", "Rahul", "Priya", "Aman", "Sneha"],
    "Age": [19, 20, 18, 21, 19],
    "Marks": [90, 85, 95, 88, 92],
    "City": ["Bhopal", "Indore", "Ujjain", "Bhopal", "Gwalior"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display Dataset
print("Dataset:")
print(df)

# Basic Information
print("\nDataset Information:")
print(df.info())

# Statistical Summary
print("\nStatistical Summary:")
print(df.describe())

# Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Average Marks
print("\nAverage Marks:", df["Marks"].mean())

# Highest Marks
print("Highest Marks:", df["Marks"].max())

# Lowest Marks
print("Lowest Marks:", df["Marks"].min())
```
